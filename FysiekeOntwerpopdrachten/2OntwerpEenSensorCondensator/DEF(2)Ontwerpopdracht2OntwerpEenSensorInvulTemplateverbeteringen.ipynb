```python
#Veerconstante bepaaldt voor in de ontwerpgrafiek: 
#uitgaande van een lineaire verband tussen stugheid van de spons en de kracht die nodig is om deze in te duwen. 

import numpy as np
import matplotlib.pyplot as plt

g = 9.81

# Afstanden in meter
afstand_0 = 1.2 / 100
afstand_1 = 1.1 / 100
afstand_2 = 1.0 / 100
afstand_3 = 0.9 / 100

# Massa's in kg
massa_0 = 0
massa_1 = 0.965 + 0.964 + 0.574
massa_2 = 0.965 + 0.964 + 0.574 + 0.4892
massa_3 = 0.965 + 0.964 + 0.574 + 0.9508 + 0.4892 + 0.4827 + 0.9738

massa = np.array([massa_1, massa_2, massa_3])
afstand = np.array([afstand_1, afstand_2, afstand_3])

# Indeuking
delta_x = afstand_0 - afstand

# Kracht
F = massa * g

# Veerconstante per meting
k = F / delta_x

# Gemiddelde veerconstante
k_gemiddeld = np.mean(k)

print("Veerconstante per meting (N/m):", k)
print("Gemiddelde veerconstante (N/m):", k_gemiddeld)
print("de gevonden waarde is veel hoger dan dat je zou verwachten, aangezien een normale huispons een veerconstante heeft tussen de 300 en 500 N/m. Deze hoge waarde kan worden verklaard doordat we relatief kleine massa's gebruiken en het aan het begin moeilijker is om de spons in te duwen, waardoor de gemeten veerconstante hoger uitvalt, maar ook omdat de spons mogelijk niet lineair is en stugger wordt naarmate deze verder wordt ingedrukt.")
```

```python
#Ontwerpgrafiek: 

epsilon_0 = 8.854187817e-12
epsilon_r = 1.5
oppervlakte = 0.00992
afstand_0 = 1.2 / 100  # 1.2 cm → meter

k = k_gemiddeld
g = 9.81

def capaciteit_plaatcondensator(oppervlakte, afstand, epsilon_r):
    return epsilon_0 * epsilon_r * oppervlakte / afstand

# Massa's van 0 tot 6 kg
massa = np.linspace(0, 6, 100)

# Bereken indeuking
delta_x = (massa * g) / k

# Nieuwe afstand
afstand = afstand_0 - delta_x

# Voorkom negatieve afstanden
afstand[afstand <= 0] = 1e-6

# Bereken capaciteit (in Farad)
Capciteit_voorspelling = capaciteit_plaatcondensator(oppervlakte, afstand, epsilon_r)

# Omzetten naar picoFarad
Capaciteit_voorspellingpF = Capciteit_voorspelling * 1e12

# Plot indeuking tegen capaciteit in pF
plt.plot(delta_x, Capaciteit_voorspellingpF)
plt.xlabel("Indeuking (m)")
plt.ylabel("Capaciteit (pF)")
plt.title("Indeuking vs Capaciteit")
plt.show()
```

```python
#Punten van de gemeten waarden: 

epsilon_0 = 8.854187817e-12
epsilon_r_spons = 1.5
oppervlakte = 0.01
afstand_0 = 1.2 / 100  
g = 9.81

gebruikte_massa  =  np.array([964, 1446.7, 2397.5, 2971.5, 3460.7, 4425.1, 5398.9, 6361.9 ]) / 1000
capaciteit_gemeten = np.array([14.7, 16.9, 23.1, 26.4, 28.9, 32.6, 36.2, 40.2 ]) * 1e-12


# Bereken indeuking
delta_x_spons = (gebruikte_massa * g) / k_gemiddeld

# Nieuwe afstand
afstand_spons = afstand_0 - delta_x_spons

# Voorkom negatieve afstanden
afstand_spons[afstand_spons <= 0] = 1e-6

# Omzetten naar picoFarad
capaciteit_gemetenpF = capaciteit_gemeten * 1e12


#Onzekerheden: 
sigma_C = 0.01  # pF
sigma_m = 0.01 / 1000  # gram → kg
sigma_x = (g / k_gemiddeld) * sigma_m

# Plot indeuking tegen capaciteit in pF
plt.plot(delta_x_spons, capaciteit_gemetenpF, 'r.', label='Gemeten waarden')
plt.errorbar(delta_x_spons, capaciteit_gemetenpF, xerr=sigma_x, yerr=sigma_C, fmt='none', ecolor='red', elinewidth=1, capsize=3, label='Meetfout')
plt.plot(delta_x, Capaciteit_voorspellingpF, label='Voorspellde waarden')
plt.xlabel("Indeuking (m)")
plt.ylabel("Capaciteit (pF)")
plt.title("Indeuking vs Capaciteit")
plt.legend()
plt.show()

 



```

```python
#Kalibratiegrafek: rekening houdend met dat de k-waarde niet constant is maar varieert afhankelijk van de indeuking en dat Arduino al een offset fout meegaf.
from scipy.optimize import curve_fit

# De ontwerpfunctie die we gaan fitten
def ontwerp_model(m, k_fit, offset):
    delta_x = (m * 9.81) / k_fit
    d = np.maximum(0.012 - delta_x, 0.0001)
    return (8.854e-12 * 1.5 * 0.01 / d) * 1e12 + offset

# Echte k en offset op basis van je rode punten
popt, _ = curve_fit(ontwerp_model, gebruikte_massa, capaciteit_gemetenpF, p0=[7000, 5])
k_echt, offset_echt = popt

# Plot deze nieuwe 'Kalibratiefit' lijn (bijv. in groen)
C_fit = ontwerp_model(massa, k_echt, offset_echt)
plt.plot(delta_x, C_fit, 'g-', label=f'Kalibratiefit (k={k_echt:.0f})')
```

Onze sensor kan nog verbetered worden door de volgende verbeteringen

## Verbetering 1: Stabiele en uniforme druk 

**Het probleem:**  
In onze huidige opstelling gebruikten we ducttape om de platen op de spons te houden en zandzakken als gewicht. Het probleem is dat zandzakken hun gewicht niet overal gelijk verdelen. Hierdoor kan de bovenste plaat aan de ene kant harder op de spons drukken dan aan de andere kant, waardoor de platen scheef komen te staan. In onze formule gaan we er echter vanuit dat de afstand overal precies gelijk is. 

**De verbetering:**  
We kunnen de opstelling verbeteren door massieve metalen gewichten te gebruiken die exact even groot zijn als de condensatorplaten. Nog beter zou een geleidingsframe zijn (bijvoorbeeld vier verticale pennen waar de bovenplaat langs glijdt). Dit dwingt de platen om altijd kaarsrecht en parallel naar beneden te bewegen, waardoor de afstand  over het hele oppervlak gelijk blijft en onze metingen veel nauwkeuriger worden. 

## Verbetering 2: Randeffecten en luchtinsluiting 

**Het probleem:**  
Ons theoretisch model is een versimpeling. In werkelijkheid buigen de elektrische veldlijnen aan de randen van de platen naar buiten. Daarnaast zorgt de ducttape ervoor dat er kleine luchtbelletjes tussen de plaat en de spons kunnen zitten. Lucht heeft een andere permittiviteit dan de spons, wat kleine afwijkingen veroorzaakt die niet in ons ontwerpmodel zijn opgenomen. 

**De verbetering:**  
We kunnen dit verbeteren door grotere condensatorplaten te gebruiken; hoe groter de platen ten opzichte van de afstand tussen de platen, hoe minder last we hebben van de buigende veldlijnen aan de randen. Daarnaast kunnen we de platen direct op de spons verlijmen met een dunne laag geleidende lijm. Dit verwijdert alle lucht tussen de plaat en de spons, waardoor de gemeten capaciteit veel beter overeenkomt met de theoretische waarde uit onze formule.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# --- 1. DATA EN CONSTANTEN ---
g = 9.81
epsilon_0 = 8.854e-12
epsilon_r = 1.5
oppervlakte = 0.01
afstand_0 = 0.012 

m_meet = np.array([964, 1446.7, 2397.5, 2971.5, 3460.7, 4425.1, 5398.9, 6361.9]) / 1000
C_meet = np.array([14.7, 16.9, 23.1, 26.4, 28.9, 32.6, 36.2, 40.2]) # pF

# --- 2. DE ONTWERPFUNCTIE (HET MODEL) ---
def sensor_model(m, k_fit, offset):
    delta_x = (m * g) / k_fit
    d = np.maximum(afstand_0 - delta_x, 0.0001)
    C_theo = (epsilon_0 * epsilon_r * oppervlakte / d) * 1e12
    return C_theo + offset

# --- 3. DE FIT UITVOEREN (VOOR DE GROENE LIJN) ---
# We zoeken de echte k en offset op basis van je metingen
popt, _ = curve_fit(sensor_model, m_meet, C_meet, p0=[7000, 5])
k_echt, offset_echt = popt

# --- 4. ALLES PLOTTEN IN ÉÉN GRAFIEK ---
m_lijn = np.linspace(0, 7, 100)
x_as_lijn = (m_lijn * g) / k_echt # We plotten tegen indeuking zoals in je plaatje

plt.figure(figsize=(10, 6))

# A. De metingen (Rood)
plt.scatter((m_meet * g) / k_echt, C_meet, color='red', label='Metingen (Kalibratie)')

# B. De Ontwerpgrafiek (Blauw - je eerste voorspelling met k_gemiddeld)
C_ontwerp = sensor_model(m_lijn, 18963, 0) # Gebruik hier je oude k-waarde
plt.plot(x_as_lijn, C_ontwerp, 'b--', label='Ontwerpgrafiek (Voorspelling k=18963)')

# C. De Kalibratiefit (Groen - je nieuwe plaatje)
C_fit = sensor_model(m_lijn, k_echt, offset_echt)
plt.plot(x_as_lijn, C_fit, 'g-', label=f'Kalibratiefit (Gevonden k={k_echt:.0f})')

plt.xlabel("Indeuking (m)")
plt.ylabel("Capaciteit (pF)")
plt.title("Sensor Analyse: Ontwerp vs. Meting")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Toepassing van de ontwerpfunctie

We hebben de natuurkundige formule van een plaatcondensator gebruikt:

Opvallend is dat de meetpunten de exponentiële kromme al heel vroeg volgen. De capaciteit verandert dus al sterk bij een kleine indrukking. Voor een touch‑sensor is dit ideaal: een lichte aanraking geeft al een duidelijk signaal.

De sensor werkt betrouwbaar en is volledig te verklaren met de natuurkunde van de vlakke plaat. De afwijking van ons oorspronkelijke ontwerp is gecorrigeerd via de kalibratiefit, waardoor we de sensor nu precies kunnen uitlezen voor touch‑toepassingen.
