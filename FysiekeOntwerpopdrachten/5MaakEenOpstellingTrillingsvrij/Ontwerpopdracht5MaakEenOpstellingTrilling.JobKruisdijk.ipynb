## Opdracht 7: Plotten maar

````python
data = {
    5.7: {
        "no_damping": [1.76, 1.4, 1.86, 1.48, 1.72],
        "damping":    [0.27, 0.19, 0.22, 0.25, 0.27]
    },
    10.2: {
        "no_damping": [0.90, 1.95, 1.8, 1.75, 1.64],
        "damping":    [0.005, 0.08, 0.09, 0.11, 0.13]
    },
    3.1: {
        "no_damping": [1.52, 1.46, 1.4, 1.94, 1.41],
        "damping":    [0.17, 0.25, 0.21, 0.16, 0.24]
    },
    2.5: {
        "no_damping": [0.55, 0.57, 0.48, 0.33, 0.79],
        "damping":    [0.15, 0.18, 0.15, 0.10, 0.25]
    }
}

frequencies = []
T_values = []
T_errors = []

for f, values in data.items():
    no_damp = np.array(values["no_damping"])
    damp = np.array(values["damping"])
    
    T = damp / no_damp
    
    frequencies.append(f)
    T_values.append(np.mean(T))
    T_errors.append(np.std(T, ddof=1))  # fout (spreiding)
frequencies = np.array(frequencies)
T_values = np.array(T_values)
T_errors = np.array(T_errors)


idx = np.argsort(frequencies)
frequencies = frequencies[idx]
T_values = T_values[idx]
T_errors = T_errors[idx]


plt.figure()

plt.errorbar(
    frequencies,
    T_values,
    yerr=T_errors,
    fmt='o',
    capsize=5,
    label="Metingen ± 1σ"
)

plt.axhline(1/3, linestyle='--', label="Eis (1/3)")

plt.xlabel("Frequentie (Hz)")
plt.ylabel("Amplitude overdracht T")
plt.title("Amplitude overdracht vs frequentie")
plt.legend()
plt.grid()

plt.show()
````
